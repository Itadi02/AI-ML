## Value counts

Get count of unique values in a particular column/Series.

````python
# what are the unique genre in imdb_df?
unique_genres = (
    df['genre']
    .dropna()
    .str.split(',')
    .explode()
    .str.strip()
    .unique()
)
print(sorted(unique_genres))
````

````python
# plotting value counts of unique genres as a bar chart
import matplotlib.pyplot as plt

genre_counts = (
    df['genre']
    .dropna()
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
)

print(genre_counts)

plt.figure()
genre_counts.plot(kind='bar')

plt.xlabel("genres")
plt.ylabel("count")
plt.title("Value Counts of Unique Genres")

plt.xticks(rotation=90)
plt.show()
````

````python
# plotting value counts of unique genres as a pie chart
plt.figure()
genre_counts.plot(kind='pie', autopct='%1.1f%%')

plt.ylabel("")  # removes default ylabel
plt.title("Distribution of Genres")

plt.show()
````

## Group By

Any groupby operation involves one of the following operations on the original object. They are −

- Splitting the Object

- Applying a function

- Combining the results

In many situations, we split the data into sets and we apply some functionality on each subset. In the apply functionality, we can perform the following operations −

- **Aggregation** − computing a summary statistic

- **Transformation** − perform some group-specific operation

- **Filtration** − discarding the data with some condition

````python
# group imdb_df by movie genres
df.groupby('genre')
df
````

````python
# get crime movies group
df.groupby('genre').get_group('Crime')
````

````python
# get mean of movie durations for each group
df.groupby('genre')['duration'].mean()
````

````python
# change duration of all movies in a particular genre to mean duration of the group
df['duration'] = df.groupby('genre')['duration'].transform('mean')
````

````python
# drop groups/genres that do not have average movie duration greater than 120.
df = df.groupby('genre').filter(lambda x: x['duration'].mean() > 120)
````
